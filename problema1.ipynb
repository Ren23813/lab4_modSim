# Ejercicio 1

Maximizar $z=120x_1+80x_2+100x_3+150x_4$

Sujeto a
$$
4x_1 + x_2 + 2x_3 + 3x_4 \leq 60\\
2x_1 + 2x_2 + x_3 +3x_4 \leq 48 \\
x_1 + x_2 + 3x_3 + 2x_4 \leq 36 \\
3x_1 + x_2 + 2x_3 + 4x_4 \leq 30 \\
x_1,x_2,x_3,x_4\geq0
$$

## Solución

```julia
using JuMP
using HiGHS
```

```julia
model = Model()

objetivo = [120, 80, 100, 150]

coeficientes = [
    4 1 2 3;
    2 2 1 3;
    1 1 3 2;
    3 1 2 4
]

@variable(model, x[1:4] >= 0)

@constraint(model, sum(x[i] * coeficientes[1, i] for i in 1:4) <= 60)
@constraint(model, sum(x[i] * coeficientes[2, i] for i in 1:4) <= 48)
@constraint(model, sum(x[i] * coeficientes[3, i] for i in 1:4) <= 36)
@constraint(model, sum(x[i] * coeficientes[4, i] for i in 1:4) <= 30)

@objective(model, Max, sum(objetivo[i] * x[i] for i in 1:4))
```

```julia
set_optimizer(model, HiGHS.Optimizer)

model
```

```julia
optimize!(model)

print("\n")
print("sol = ", value.(x), "\n")
print("max = ", objective_value(model))
```
